# Feature Engineering – Fraud Detection

This notebook creates new features from raw transaction data to improve
fraud detection performance.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("seaborn-v0_8")
sns.set_context("notebook")
```

```python
df = pd.read_csv("../data/raw/Fraud_Data.csv")
df['signup_time'] = pd.to_datetime(df['signup_time'])
df['purchase_time'] = pd.to_datetime(df['purchase_time'])
```

```python
df['hour_of_day'] = df['purchase_time'].dt.hour
df['day_of_week'] = df['purchase_time'].dt.dayofweek

df['time_since_signup'] = (
    df['purchase_time'] - df['signup_time']
).dt.total_seconds() / 3600
```

```python
df['purchase_time'] = pd.to_datetime(df['purchase_time'])
df = df.sort_values(['user_id', 'purchase_time'])

counts = (
    df.set_index('purchase_time')
      .groupby('user_id')['user_id']
      .rolling('24H')
      .count()
      .reset_index(level=0, drop=True)
)

df['txn_count_24h'] = counts.values
```

```python
df.to_csv("../data/processed/fraud_processed.csv", index=False)
```
